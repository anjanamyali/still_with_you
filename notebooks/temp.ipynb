```python
import base64
import hashlib
import json
from pathlib import Path

import requests


def clone_voice(
    audio_file: str,
    display_name: str,
    transcription: str,
    metadata_save_path: Path,
    lang_code: str = "EN_US",
    api_key: str = None,
) -> str:
    """
    Clone a voice and return voice_id.
    """

    if metadata_save_path.exists():
        voice_info = json.loads(metadata_save_path.read_text())
        voice_id = voice_info.get("voice_id")

        if voice_id:
            return voice_id


    headers = {
        "Authorization": f"Basic {api_key}",
        "Content-Type": "application/json",
    }

    with open(audio_file, "rb") as f:
        audio_b64 = base64.b64encode(f.read()).decode("utf-8")

    payload = {
        "displayName": display_name,
        "langCode": lang_code,
        "voiceSamples": [
            {
                "audioData": audio_b64,
                # API currently expects transcription
                "transcription": transcription,
            }
        ],
        "description": "",
        "tags": [],
        "audioProcessingConfig": {
            "removeBackgroundNoise": True
        },
    }

    r = requests.post(
        "https://api.inworld.ai/voices/v1/voices:clone",
        headers=headers,
        json=payload,
        timeout=300,
    )

    r.raise_for_status()

    data = r.json()
    voice_id = data["voice"]["voiceId"]

    if metadata_save_path.exists():
        voice_info = json.loads(metadata_save_path.read_text())
    else:
        voice_info = {
            "display_name": display_name,
        }

    voice_info["voice_id"] = voice_id
    metadata_save_path.write_text(json.dumps(voice_info, indent=2))

    return voice_id

def synthesize(
    text: str,
    voice_id: str,
    wav_save_path: Path, # .wav file save path
    model_id: str = "inworld-tts-1.5-mini",
    sample_rate: int = 22050,
) -> bytes:
    """
    Generate speech.
    Cached by voice_id + transcript.
    Returns WAV bytes.
    """

    json_save_path = Path(wav_save_path.replace(".wav", ".json"))

    if wav_save_path.exists():
        return wav_save_path.read_bytes()

    payload = {
        "text": text,
        "voiceId": voice_id,
        "modelId": model_id,
        "audioConfig": {
            "audioEncoding": "LINEAR16",
            "sampleRateHertz": sample_rate,
        },
        "deliveryMode": "BALANCED",
        "applyTextNormalization": "ON",
    }

    r = requests.post(
        "https://api.inworld.ai/tts/v1/voice",
        headers=headers,
        json=payload,
        timeout=300,
    )

    r.raise_for_status()

    data = r.json()

    wav_bytes = base64.b64decode(data["audioContent"])

    wav_save_path.write_bytes(wav_bytes)

    meta = {
        "text": text,
        "voice_id": voice_id,
        "usage": data.get("usage", {}),
    }

    json_save_path.write_text(
        json.dumps(meta, indent=2)
    )

    return wav_bytes
```
